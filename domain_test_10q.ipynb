```python
#two sum
nums=[1,2,3,4,5]
target=9
l1=[]
for i in range(len(nums)):
  for j in range(i+1,len(nums)):
    if nums[i]+nums[j]==target:
      l1.append(i)
      l1.append(j)
print(l1)
```

```python
#two sum another
nums=[1,2,3,4,5]
target=9
l1=[]
left=0
right=len(nums)-1
while left<right:
  if nums[left]+nums[right]==target:
    l1.append(left)
    l1.append(right)
    right-=1
    left+=1
  elif nums[left]+nums[right]>target:
    right-=1
  else:
    left+=1
print(l1)
```

```python
#2nd largest distinct
nums= [10, 20, 40, 30, 40 ,-5,-4]
large=0
second=0
for i in range(len(nums)):
  if nums[i]>large:
    second=large
    large=nums[i]
  elif nums[i]>second and nums[i]!=large:
    second=nums[i]
print(second)
```

```python
#palindrome sentence
s = "A man, a plan, a canal: Panama"
t=[]
for i in s:
  if i.isalnum():
    t.append(i.lower())
u="".join(t)
print(u==u[::-1])
```
